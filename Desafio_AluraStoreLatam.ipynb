### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
print(tienda.columns)

```

Comenzamos Conociendo cuales eran nuestras "celdas" para trabajar

```python


# Sumar la facturación (Columna de Precio) de cada tienda
facturacion_tienda1 = tienda['Precio'].sum()
facturacion_tienda2 = tienda2['Precio'].sum()
facturacion_tienda3 = tienda3['Precio'].sum()
facturacion_tienda4 = tienda4['Precio'].sum()

# Con esto vamos a mostrar la facturación de cada tienda
print(f"Facturación Tienda 1: ${facturacion_tienda1:,.2f}")
print(f"Facturación Tienda 2: ${facturacion_tienda2:,.2f}")
print(f"Facturación Tienda 3: ${facturacion_tienda3:,.2f}")
print(f"Facturación Tienda 4: ${facturacion_tienda4:,.2f}")

# Facturación total de todas las tiendas
facturacion_total = facturacion_tienda1 + facturacion_tienda2 + facturacion_tienda3 + facturacion_tienda4
print(f"Facturación Total: ${facturacion_total:,.2f}")

```

# 2. Ventas por categoría

```python
#Agrupar las ventas por la columna (categoria del producto)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Agrupar por categoría y sumar las ventas
ventas_por_categoria = tienda.groupby('Categoría del Producto')['Precio'].sum().reset_index()

# Ordenar las categorías de mayor a menor
ventas_por_categoria = ventas_por_categoria.sort_values('Precio', ascending=False)

# Mostrar el resultado
print(ventas_por_categoria)

# Graficar las ventas por categoría
plt.figure(figsize=(10,6))
sns.barplot(x='Precio', y='Categoría del Producto', data=ventas_por_categoria, hue='Categoría del Producto', palette='viridis', legend=False)

# Mejorar el gráfico

# Mejorar el gráfico
plt.title('Ventas por Categoría de Producto', fontsize=16)
plt.xlabel('Ventas Totales ($)', fontsize=12)
plt.ylabel('Categoría de Producto', fontsize=12)
plt.tight_layout()

# Mostrar
plt.show()
```

# 3. Calificación promedio de la tienda


```python
# Calcular la calificación promedio de cada tienda
calificacion_tienda1 = tienda['Calificación'].mean()
calificacion_tienda2 = tienda2['Calificación'].mean()
calificacion_tienda3 = tienda3['Calificación'].mean()
calificacion_tienda4 = tienda4['Calificación'].mean()

# Mostrar el resultado
print(f"Calificación promedio Tienda 1: {calificacion_tienda1:.2f}")
print(f"Calificación promedio Tienda 2: {calificacion_tienda2:.2f}")
print(f"Calificación promedio Tienda 3: {calificacion_tienda3:.2f}")
print(f"Calificación promedio Tienda 4: {calificacion_tienda4:.2f}")

# Crear un DataFrame para graficar las calificaciones
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
calificaciones = [calificacion_tienda1, calificacion_tienda2, calificacion_tienda3, calificacion_tienda4]

# Graficar las calificaciones promedio por tienda
plt.figure(figsize=(10,6))
sns.barplot(x='Tienda', y='Calificación Promedio', data=pd.DataFrame({'Tienda': tiendas, 'Calificación Promedio': calificaciones}), hue='Tienda', palette='Blues_d', legend=False)


# Mejorar el gráfico
plt.title('Calificación Promedio por Tienda', fontsize=16)
plt.xlabel('Tienda', fontsize=12)
plt.ylabel('Calificación Promedio', fontsize=12)
plt.tight_layout()


# Mostrar
plt.show()

```

# 4. Productos más y menos vendidos

```python
# Agrupar por producto y sumar la cantidad de ventas
productos_ventas = tienda.groupby('Producto')['Cantidad de cuotas'].sum().reset_index()

# Ordenar los productos de mayor a menor
productos_ventas = productos_ventas.sort_values('Cantidad de cuotas', ascending=False)

# Mostrar los productos más vendidos y los menos vendidos
# Los 10 productos más vendidos
print("Productos más vendidos:\n")
print(productos_ventas.head(10))

# Los 10 productos menos vendidos
print("\nProductos menos vendidos:\n")
print(productos_ventas.tail(10))

# Graficar los productos más vendidos
plt.figure(figsize=(12,8))
sns.barplot(x='Cantidad de cuotas', y='Producto', data=productos_ventas.head(10), hue='Producto', palette='viridis', legend=False)

# Mejorar el gráfico
plt.title('Productos Más Vendidos', fontsize=16)
plt.xlabel('Cantidad de Unidades Vendidas', fontsize=12)
plt.ylabel('Producto', fontsize=12)
plt.tight_layout()

# Mostrar
plt.show()

```

# 5. Envío promedio por tienda

```python
tienda['Tienda'] = 'Tienda 1'
tienda2['Tienda'] = 'Tienda 2'
tienda3['Tienda'] = 'Tienda 3'
tienda4['Tienda'] = 'Tienda 4'

# Unir todos los datos
datos = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)

# Agrupar por tienda y calcular el costo promedio de envío
envio_promedio = datos.groupby('Tienda')['Costo de envío'].mean().reset_index()

# Redondear a 2 decimales
envio_promedio['Costo de envío'] = envio_promedio['Costo de envío'].round(2)

# Ordenar por costo de envío de mayor a menor
envio_promedio = envio_promedio.sort_values('Costo de envío', ascending=False)

# Mostrar los costos promedio de envío
print(envio_promedio)

# Graficar el costo promedio de envío por tienda
plt.figure(figsize=(10,6))
grafico = sns.barplot(x='Costo de envío', y='Tienda', data=envio_promedio, hue='Tienda', palette='coolwarm', legend=False)

# Agregar valores encima de las barras
for container in grafico.containers:
    grafico.bar_label(container, fmt='%.2f', label_type='edge', padding=3)

# Mejorar el gráfico
plt.title('Costo Promedio de Envío por Tienda', fontsize=16)
plt.xlabel('Costo de Envío Promedio (USD)', fontsize=12)
plt.ylabel('Tienda', fontsize=12)
plt.tight_layout()

plt.show()



```
